# Czy zawodnik X nadaje się na wahadłowego?

Na początku określiłbym, czego dokładnie oczekujemy od wahadłowego w naszym systemie. Sama nazwa pozycji nie wystarczy, bo w zależności od modelu gry wahadłowy może mieć inne zadania. Jeśli gramy 1-3-5-2, interesowałby mnie zawodnik, który daje szerokość i progresję w ataku, ale jednocześnie potrafi szybko wrócić po stracie i bronić bocznego sektora.

## Jak bym do tego podszedł

Mając dane StatsBomb z 20 meczów, najpierw sprawdziłbym, na jakich pozycjach zawodnik rzeczywiście grał i w jakim kontekście. Inaczej interpretowałbym liczby skrzydłowego, inaczej bocznego obrońcy, a inaczej zawodnika, który już grał jako wahadłowy.

W ofensywie patrzyłbym przede wszystkim na prowadzenia z piłką, progresywne podania, wejścia w ostatnią tercję, podania w pole karne, dośrodkowania, cutbacki, tworzenie sytuacji.

Bez piłki sprawdziłbym pressing, odbiory, przechwyty i pojedynki, szczególnie w bocznym sektorze. Interesowałoby mnie też, jak zawodnik reaguje po stracie i czy potrafi szybko odbudować pozycję.

Należałoby również porównać zawodnika z innymi graczami z podobnej roli, najlepiej z tej samej ligi i zespołów o zbliżonym stylu gry.

Na końcu użyłbym danych jako filtra do analizy wideo. Jeśli liczby pokazują np. dobrą progresję, ale słabszą aktywność defensywną, to właśnie te elementy chciałbym dokładniej obejrzeć w meczu.

## Gdzie dane mogą zawieść

Największe ograniczenie StatsBomb jest takie, że event data dobrze pokazują akcje przy piłce, ale dużo słabiej rzeczy bez piłki. Nie zobaczymy dobrze ustawienia, jakości powrotów, ruchu bez piłki czy tego, czy zawodnik prawidłowo zamyka przestrzeń. Dlatego na podstawie samych danych nie dawałbym ostatecznej odpowiedzi „tak” lub „nie”. Traktowałbym je jako pierwszy etap oceny, a decyzję opierał na połączeniu danych z analizą wideo.
